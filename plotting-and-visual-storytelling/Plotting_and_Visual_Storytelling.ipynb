```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
print(tips.head())
```

```python
#Task 1 — Your First Matplotlib Chart

tip_sum = tips.groupby("day")["tip"].sum()
plt.figure(figsize=(8, 5))
plt.bar(tip_sum.index, tip_sum.values)

plt.title("Total Tips Collected per Day")
plt.xlabel("Day")
plt.ylabel("Total Tip Amount")

plt.show()
```

```python
#Task 2 — Seaborn Histogram

plt.figure(figsize=(8, 5))
sns.histplot(data=tips, x="tip", bins=20, kde=True)
plt.title("Distribution of Tip Amounts")

plt.show()
```

```python
'''Explanation
The histogram shows that most tip amounts fall within a lower range, typically between 2 and 4.
The distribution appears slightly right-skewed, indicating fewer high-value tips compared to smaller ones.
The KDE curve helps visualize the overall trend more smoothly, making it easier
to identify the peak and spread of the data.'''
```

```python
#Task 3 — Seaborn Scatter Plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="day")

plt.title("Total Bill vs Tip Amount")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
'''Explanation
The scatter plot shows a clear positive relationship between the total bill and the tip amount,
meaning higher bills generally lead to higher tips. The color distribution across days suggests
this pattern is consistent regardless of the day of the week.'''
```
