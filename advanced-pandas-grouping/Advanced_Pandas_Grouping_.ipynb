```python
import pandas as pd
import random

random.seed(42)

# Define data parameters
regions = ['North', 'South', 'East', 'West']
categories = ['Electronics', 'Clothing', 'Home & Garden', 'Sports', 'Books']
salespersons = ['Alice', 'Bob', 'Carol', 'David', 'Emma', 'Frank']

# Generate 200 sales transactions
data = {
    'transaction_id': range(1001, 1201),
    'region': [random.choice(regions) for _ in range(200)],
    'category': [random.choice(categories) for _ in range(200)],
    'salesperson': [random.choice(salespersons) for _ in range(200)],
    'sales_amount': [round(random.uniform(50, 5000), 2) for _ in range(200)],
    'customer_id': [random.randint(5000, 5100) for _ in range(200)]
}

df = pd.DataFrame(data)
print(df.head(10))
print(f"\nDataset shape: {df.shape}")
```

```python
#===Task 1: Basic Grouping and Single Aggregations===
#Calculate total sales amount for each region
region_sales = df.groupby('region')['sales_amount'].sum()
print(f"Sales on each region: {region_sales}")
```

```python
#Count the number of transactions for each product category
category_counts = df.groupby('category')['transaction_id'].count()
print(f"Number of transactions for each product category:\n {category_counts}")
```

```python
#Calculate the average sales amount per salesperson
salesperson_avg_sales = df.groupby('salesperson')['sales_amount'].mean()
print(f"Average sales amount per salesperson:\n {salesperson_avg_sales}")
```

```python
#For each result, use reset_index() to convert the result into a clean DataFrame
region_sales_with_index = region_sales.reset_index()
category_counts_with_index = category_counts.reset_index()
salesperson_avg_sales_with_index = salesperson_avg_sales.reset_index()
print(f"Sales on each region with index:\n {region_sales_with_index}")
print(f"Number of transactions for each product category with index:\n {category_counts_with_index}")
print(f"Average sales amount per salesperson with index:\n {salesperson_avg_sales_with_index}")
```

```python
#Sort the regional sales results in descending order
region_sales_sorted = region_sales.sort_values(ascending=False)
print(f"Regional sales results in top performance order:\n {region_sales_sorted}")
```

```python
#===Task 2: Multi-Column Grouping and Multiple Aggregations===
#Multi-Column Grouping and Multiple Aggregations
result= df.groupby(['region', 'category'])['sales_amount'].sum().reset_index()
print(f"Multi-Column Grouping and sales amount aggregations:\n {result}")
```

```python
#For each salesperson, calculate three metrics simultaneously using the agg() method:
result= df.groupby('salesperson')['sales_amount'].agg(['sum', 'mean', 'count']).reset_index()
print(f"All three metrics for each salesperson\n: {result}")
```

```python
#Sort the salesperson results by total sales in descending order to identify the top performer
top_performer = df.groupby('salesperson')['sales_amount'].sum().sort_values(ascending=False).head(1)
print(f"Top perfoeming salesman:\n {top_performer}")
```

```python
#Use .idxmax() on the grouped category sales to find which category has the maximum total revenue
category_with_max_revenue = df.groupby('category')['sales_amount'].sum().idxmax()
print(f"Category with most revenue: {category_with_max_revenue}")
```

```python
#===Task 3: Custom Aggregation and Complete Sales Report===
#Define a custom aggregation function that calculates the sales range (max - min) for each group:
def sales_range(x):
    return x.max() - x.min()
```

```python
#Apply this custom function along with standard aggregations to analyze sales by region:
result= df.groupby('region')['sales_amount'].agg(['sum', 'mean', 'min', 'max', sales_range]).reset_index()
print(f"custom functions with std aggregations: \n {result}")
```

```python
#Create a final summary report that shows for each region:

final_report = df.groupby('region').agg({
    'sales_amount': ['sum', 'mean'],
    'transaction_id': 'count'
}).reset_index()


print("Final Summary Report by Region:")
print(final_report)
```

```python
#Explanation
"""When you perform a single aggregation, like df.groupby('region')['sales_amount'].sum(), the result is a pandas Series with a single level of labels for the data. The column holding the aggregated values will simply be named after the aggregated column (e.g., sales_amount).
However, when you use dictionary-based aggregation with multiple aggregations for the same column,
pandas creates a MultiIndex (or hierarchical index) for the columns. This means the column names will have two levels:

Top level: The name of the original column you applied the aggregation to (e.g., sales_amount, transaction_id).
Second level: The name of the aggregation function applied (e.g., sum, mean, count).
So, instead of having columns like sales_amount_sum and sales_amount_mean, you'll have a structure where sales_amount is a top-level column, and sum and mean are sub-columns under it. Similarly, transaction_id would be a top-level column with count as a sub-column."""
```
