```python
%%sql
CREATE TABLE employees (
    employee_id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary INTEGER,
    hire_date DATE
);
```

```python
import sqlite3
import random
from datetime import datetime, timedelta

conn = sqlite3.connect('company.db')
cursor = conn.cursor()

departments = ['Engineering', 'Sales', 'Marketing', 'Finance', 'HR', 'Operations']

data = []
for i in range(1, 101):
    name = f'Employee_{i}'
    dept = random.choice(departments)
    salary = random.randint(40000, 100000)
    hire_date = datetime(2015,1,1) + timedelta(days=random.randint(0, 3000))

    data.append((i, name, dept, salary, str(hire_date.date())))

cursor.executemany(
    "INSERT INTO employees (employee_id, name, department, salary, hire_date) VALUES (?, ?, ?, ?, ?)",
    data
)

conn.commit()
conn.close()
```

```python
#Task 1: Basic Querying
#Select all columns for employees in the Marketing department
%%sql
SELECT * FROM employees WHERE department = 'Marketing';
```

```python
#Display only name, department, and salary for employees earning more than $90,000
%%sql
SELECT name,department,salary FROM employees WHERE salary > 90000;
```

```python
#Find employees in either Sales OR Finance departments
%%sql
SELECT * from employees WHERE department IN ('Sales','Finance');
SELECT * from employees WHERE department = 'Sales' OR department = 'Finance';
```

```python
#Task 2: Sorting and Limiting
#The top 5 highest-paid employees (sort by salary descending, limit to 5)
%%sql
SELECT * FROM employees ORDER BY salary DESC LIMIT 5;
```

```python
#The 3 most recently hired employees in Engineering department (sort by hire_date descending)
%%sql
SELECT * FROM employees
WHERE department='Engineering'
ORDER BY hire_date desc
LIMIT 3;
```

```python
#The oldest employee (by hire_date) who earns more than $70,000 and is NOT in HR department
%%sql
SELECT * FROM employees
WHERE salary > 70000 AND department != 'HR'
ORDER BY hire_date ASC
LIMIT 1;
```

```python
#Task 3: Calculated Fields and Aliases
#Calculate monthly salary for all employees:
%%sql
SELECT name, salary, salary/12 AS monthly_pay FROM employees
WHERE salary > 60000;
```

```python
#Create a report for the Finance department
%%sql
SELECT name AS employee_name, salary AS annual_salary, salary/12 AS monthly_salary
FROM employees
ORDER BY salary desc;
```
