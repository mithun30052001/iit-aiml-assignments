```python
data = {
    'Gender': ['Male', 'Female', 'Female', 'Male', 'Female'],
    'City': ['Mumbai', 'Delhi', 'Chennai', 'Mumbai', 'Delhi'],
    'Size': ['Small', 'Large', 'Medium', 'Large', 'Small'],
    'Age': [25, 45, 32, 60, 28],
    'Fare': [15, 300, 85, 450, 20]
}
```

```python
#Task 1 — Categorical Encoding

import pandas as pd
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame(data)

#Gender → Label Encoding
le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])

#City → One-Hot Encoding (drop one column to avoid the dummy variable trap)
df = pd.get_dummies(df, columns=['City'], drop_first=True)

#Size → Label Encoding with the correct order: Small=0, Medium=1, Large=2
size_mapping = {'Small': 0, 'Medium': 1, 'Large': 2}
df['Size'] = df['Size'].map(size_mapping)

print(f"Resultant after categorical encoding\n:{df}")
```

```python
#Task 2 — Feature Scaling

from sklearn.preprocessing import RobustScaler
scaler = RobustScaler()

df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])
print(f"Resultant after scaling:\n {df}")
```

```python
'''Explanation
RobustScaler is used because Fare has an outlier (450).
It scales data using median and IQR, making it less sensitive to extreme values,
unlike StandardScaler which can be skewed by outliers.'''
```
